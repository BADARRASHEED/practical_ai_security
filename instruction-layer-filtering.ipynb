# Instruction layer filtering

This Python script filters user instructions to block harmful, malicious, or restricted commands before they are processed. It consists of three main functions and an interactive loop.

We import re (regex) again to help identify restricted keywords in user inputs.

```python
import re
```

This function defines a simple instruction filtering layer designed to block unsafe or overly complex user commands before they are executed. It works by first checking the input text against a list of restricted keywords—such as “hack,” “delete database,” “malware,” or “attack”—and rejects the instruction if any are found, ensuring malicious intent is not processed further. It also applies a length constraint (max_length = 200) to prevent excessively long or detailed inputs, which could be a sign of prompt injection or system misuse. If neither condition is triggered, the function returns True, meaning the instruction is considered valid and safe. This kind of filtering is a common safeguard in AI and security systems to enforce input sanitisation and mitigate harmful behaviour.

```python
def instruction_layer_filter(instruction):
    """
    Filters user instructions to block harmful or restricted commands.
    :param instruction: The user-provided instruction
    :return: True if the instruction is valid and safe, otherwise False
    """
    # Restricted keywords (e.g., malicious or harmful instructions)
    restricted_keywords = ["hack", "delete database", "malware", "attack"]

    # Complexity limit (e.g., too long or overly detailed commands)
    max_length = 200

    # Check for restricted keywords
    for keyword in restricted_keywords:
        if re.search(rf"\b{keyword}\b", instruction, re.IGNORECASE):
            print(f"Instruction blocked: Contains restricted keyword '{keyword}'.")
            return False

    # Check for instruction length
    if len(instruction) > max_length:
        print("Instruction blocked: Too long or complex.")
        return False

    # Passed all filters
    return True
```

This script ties together the filtering logic with a simple interactive demo for testing instructions. The process_instruction function first checks whether a user’s input passes through the instruction_layer_filter. If the input is safe, it simulates carrying out the command by printing and returning a confirmation message; if not, it returns a standard warning that the instruction has been blocked due to safety guidelines. The __main__ block then creates a continuous loop where users can type instructions in the terminal. Each instruction is filtered and processed in real time, with "exit" serving as a clean way to break the loop. This setup illustrates how an instruction filtering layer can be integrated into a user-facing system to ensure only valid, safe commands are accepted.

```python
def process_instruction(instruction):
    """
    Processes a user instruction if it passes the instruction layer filter.
    :param instruction: The user-provided instruction
    :return: A response or an error message
    """
    if instruction_layer_filter(instruction):
        # Simulate processing the instruction
        print(f"Processing instruction: '{instruction}'")
        return f"Instruction processed: {instruction}"
    else:
        return "Your instruction was blocked for violating safety guidelines."


# Interactive loop
if __name__ == "__main__":
    print("Instruction Layer Filtering Interactive Demo")
    print("Type 'exit' to quit.\n")

    while True:
        # Get user input
        user_instruction = input("Enter your instruction: ").strip()

        # Exit condition
        if user_instruction.lower() == "exit":
            print("Exiting the demo. Goodbye!")
            break

        # Process the instruction
        response = process_instruction(user_instruction)
        print(response)
        print()
```
