# Input validation

This Python script collects user input (name, age, and email), validates it using regular expressions (regex), and ensures it meets specified constraints before proceeding.

The library re (Regular Expressions) is used for pattern matching to validate input.

```python
import re
```

This function provides a basic input validation routine for three common user fields: name, age, and email. It enforces clear constraints to prevent malformed or unsafe data. For the name, it requires only alphabetic characters and spaces, with a length between 1 and 50, ensuring no numbers or special characters slip through. For age, it checks that the input is numeric and falls within a realistic range of 1 to 120. Finally, it uses a simple regular expression to verify that the email address follows the conventional username@domain.extension format. If any of these checks fail, the function raises a ValueError with a clear explanation; otherwise, it returns True. This kind of validation helps maintain data integrity and protects applications from errors or misuse.

```python
def validate_input(name, age, email):
    """
    Validates user input for name, age, and email.
    :param name: User's name
    :param age: User's age
    :param email: User's email address
    :return: True if all inputs are valid, else raises a ValueError
    """

    # Validate name (only alphabets and spaces, 1-50 characters)
    if not re.match(r"^[a-zA-Z\s]{1,50}$", name):
        raise ValueError("Invalid name: Name must only contain letters and spaces, and be 1-50 characters long.")

    # Validate age (integer between 1 and 120)
    if not age.isdigit() or not (1 <= int(age) <= 120):
        raise ValueError("Invalid age: Age must be a number between 1 and 120.")

    # Validate email (basic regex for email format)
    if not re.match(r"^[\w\.-]+@[\w\.-]+\.\w+$", email):
        raise ValueError("Invalid email: Please provide a valid email address.")

    return True
```

This script demonstrates how to integrate the validate_input function into a simple interactive program. When run, it welcomes the user and prompts them to enter their name, age, and email address. Each input is stripped of leading/trailing spaces and then passed into the validation function. If all fields are valid, the program confirms success and could continue to process or store the data. If any input fails validation, a ValueError is raised and caught, and the program prints a clear error message explaining what went wrong. This structure ensures a user-friendly interaction while maintaining strict data integrity and safety.

```python
def main():
    print("Welcome to the User Input Validation Demo!")

    try:
        # Collect input from the user
        name = input("Enter your name: ").strip()
        age = input("Enter your age: ").strip()
        email = input("Enter your email: ").strip()

        # Validate the input
        if validate_input(name, age, email):
            print("All inputs are valid! Proceeding...")
            # Here, you can save the data or process it further
    except ValueError as e:
        print(f"Input validation error: {e}")


if __name__ == "__main__":
    main()
```
